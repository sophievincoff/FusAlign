### Independent FASTA check

Bypass `IDMapAnalyzer` and scan both FASTA files for accessions only (headers). Confirm that blank-row UniProt IDs never appear as a base accession (`ENTRY` or `ENTRY-N`) in either FASTA.

```python
HEADER_ACC_RE = re.compile(r"^>(?:sp|tr)\|([A-Za-z0-9]+(?:-\d+)?)\|")

def collect_fasta_accessions(paths):
    """Return sets of full accessions and base IDs found in FASTA headers only."""
    full_accs = set()
    base_ids = set()
    n_headers = 0
    for path in paths:
        with open(path, encoding="utf-8") as handle:
            for line in handle:
                if not line.startswith(">"):
                    continue
                n_headers += 1
                match = HEADER_ACC_RE.match(line)
                if not match:
                    raise ValueError(f"Unrecognized FASTA header in {path}: {line.strip()}")
                acc = match.group(1)
                base, _isoform = parse_uniprot_accession(acc)
                full_accs.add(acc)
                base_ids.add(base)
    return full_accs, base_ids, n_headers

fasta_full_accs, fasta_base_ids, n_fasta_headers = collect_fasta_accessions(fasta_paths)
print(f"FASTA headers scanned: {n_fasta_headers:,}")
print(f"Unique FASTA accessions: {len(fasta_full_accs):,}")
print(f"Unique FASTA base IDs: {len(fasta_base_ids):,}")

blank_ids = set(blank_df["UniProtID"].astype(str))
found_in_fasta = sorted(blank_ids & fasta_base_ids)
missing_as_expected = sorted(blank_ids - fasta_base_ids)

print(f"Blank UniProtIDs: {len(blank_ids):,}")
print(f"  absent from both FASTAs (expected): {len(missing_as_expected):,}")
print(f"  unexpectedly present in a FASTA: {len(found_in_fasta):,}")

if found_in_fasta:
    surprise = blank_df[blank_df["UniProtID"].isin(found_in_fasta)].copy()
    surprise["fasta_accessions"] = surprise["UniProtID"].map(
        lambda uid: sorted(
            acc for acc in fasta_full_accs
            if acc == uid or acc.startswith(uid + "-")
        )
    )
    print("ERROR: these blank-sequence UniProtIDs do appear in FASTA headers:")
    display(surprise.head(20))
    raise AssertionError(
        f"{len(found_in_fasta)} blank UniProtIDs were found in FASTA headers; "
        "IDMapAnalyzer join likely dropped sequences incorrectly."
    )

print("Independent check passed: no blank-sequence UniProtID appears in either FASTA.")
display(blank_df[["Query", "UniProtID", "Reviewed", "Entry Name", "Length"]].head(20))
```

```python
# make a list of all the IDs that do not have sequences
no_seq_ids = blank_df["UniProtID"].tolist()
# prepare them for resubmit to UniProt ID mapping
with open(os.path.join(uniprot_dir, "uniprot_requery_3.txt"), "w") as f:
    for id in no_seq_ids:
        f.write(f"{id}\n")
```
